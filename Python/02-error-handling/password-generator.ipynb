```python
import string
import secrets
from IPython.display import clear_output

class MinimumCharsError(Exception):
    pass

def generate_password(length):
    ## Starts an empty list for the password
    choice = []

    ## Gets and stores the required characters
    numbers = string.digits
    letters = string.ascii_letters
    special = string.punctuation
    general_chars = numbers + letters + special

    required_chars = [numbers, letters, special]

    ## Enters the loop to select the characters    
    for char in required_chars:
        choice.append(secrets.choice(char))

    while len(choice) < length:
        choice.append(secrets.choice(general_chars))

    ## Randomizes the output
    secrets.SystemRandom().shuffle(choice)

    ## Transforms the output into a string
    choice = ''.join(choice) 

    return choice

attempts = 0

while True:
    clear_output(wait=True)
    try:
        length = int(input('Enter the password length: '))
        if length < 4:
            raise MinimumCharsError('The length must be at least 4 characters. Try again...')
    except ValueError:
        print('Error: The length must be given as an integer! Try again...')
    except MinimumCharsError as e:
        print(f'Error: {e}')
    else:
        password = generate_password(length)
        print(f'The generated password is: {password}')
        break
    finally:
        attempts += 1
        print(f'A total of {attempts} attempts were made...')
```
